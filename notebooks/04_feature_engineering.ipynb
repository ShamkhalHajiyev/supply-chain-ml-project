Transform preprocessed data into ML-ready features.

```python
# ============================================================================
# SETUP
# ============================================================================
import sys
import warnings
warnings.filterwarnings('ignore')
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data.preprocess import load_and_preprocess
from src.features.build_features import FeatureEngineer, build_features_pipeline

plt.style.use('seaborn-v0_8-whitegrid')
print("✅ Setup complete!")
```

```python
# Load preprocessed data
df = load_and_preprocess()
print(f"Input: {df.shape}")
```

## 4.1 Build Features


```python
# Run feature engineering pipeline
X, y = build_features_pipeline(df)

print("\n" + "=" * 70)
print("FEATURE ENGINEERING COMPLETE")
print("=" * 70)
print(f"\n📊 Feature matrix: {X.shape}")
print(f"📊 Target: {y.shape}")
print(f"\n📋 Features created ({X.shape[1]}):")
for i, col in enumerate(X.columns[:15]):
    print(f"   {i+1:2d}. {col}")
if X.shape[1] > 15:
    print(f"   ... and {X.shape[1]-15} more")
```

## 4.2 Feature Correlation with Target


```python
# Feature correlations with target
correlations = X.corrwith(y).sort_values(key=abs, ascending=False)

fig, ax = plt.subplots(figsize=(10, 8))
correlations.head(20).plot(kind='barh', ax=ax, color='steelblue')
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_xlabel('Correlation with Late Delivery')
ax.set_title('Top 20 Features by Correlation', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("\nTop 10 correlated features:")
print(correlations.head(10))

print("\n💡 INTERPRETATION:")
print("   • Shipping-related features have highest predictive power")
print("   • Days for shipment (scheduled) strongly negatively correlated")
```

```python
print("\n✅ Feature engineering complete!")
print(f"   • {X.shape[1]} features created for {X.shape[0]:,} samples")
print("\n➡️ Next: Run 05_classification_modeling.ipynb")
```
