# Step 4: ML Model Training

## Overview
This notebook covers training classification models for late delivery prediction.

### Models Trained
1. **Logistic Regression**: Baseline linear model
2. **Random Forest**: Ensemble of decision trees
3. **Gradient Boosting**: Sequential boosting approach

### Evaluation Metrics
- Accuracy, Precision, Recall, F1 Score
- ROC-AUC Score
- Confusion Matrix
- Cross-Validation Scores


```python
import sys
import warnings
warnings.filterwarnings('ignore')
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

from src.data.preprocess import load_and_preprocess
from src.features.build_features import build_features_pipeline
from src.models.train_ml import SupplyChainClassifier

plt.style.use('seaborn-v0_8-darkgrid')
print("✅ Setup complete!")
```

```python
# Load and prepare data
df = load_and_preprocess()
X, y = build_features_pipeline(df)
print(f"Feature matrix: {X.shape}")
print(f"Target distribution: {y.value_counts().to_dict()}")
```

```python
# Initialize classifier and train
classifier = SupplyChainClassifier(random_state=42)
classifier.initialize_models()

# Split data
X_train, X_test, y_train, y_test = classifier.split_data(X, y, test_size=0.2)

# Train all models
classifier.train_all_models(X_train, y_train, X_test, y_test)
```

```python
# Feature importance
importance_df = classifier.get_feature_importance(X.columns.tolist(), top_n=15)

# Visualize
fig, ax = plt.subplots(figsize=(10, 8))
importance_df.plot(x='feature', y='importance', kind='barh', ax=ax, legend=False)
ax.set_xlabel('Importance')
ax.set_title(f'Top 15 Features ({classifier.best_model_name})', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Confusion matrix for best model
y_pred = classifier.best_model.predict(X_test)

fig, ax = plt.subplots(figsize=(8, 6))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred,
    display_labels=['On-time', 'Late'],
    cmap='Blues', ax=ax
)
ax.set_title('Confusion Matrix - Late Delivery Prediction', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Save models
classifier.save_models()

# Generate report
classifier.generate_report()
```

## Key Interpretation

### Model Comparison:
- **Logistic Regression**: Fast, interpretable baseline
- **Random Forest**: Handles non-linear relationships
- **Gradient Boosting**: Often highest accuracy

### Key Metrics:
- **F1 Score**: Balance of precision and recall
- **ROC-AUC**: Model discrimination ability
- **CV Score**: Generalization performance
