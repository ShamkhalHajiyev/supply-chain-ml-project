# Step 5: LSTM Model Training

## Overview
This notebook covers training LSTM neural networks for demand forecasting.

### LSTM Architecture
- **Input Layer**: Sequence of historical data
- **LSTM Layers**: 2 layers with dropout
- **Output Layer**: Predicted demand value

### Key Concepts
- **Sequence Length**: 30 days of historical data
- **Normalization**: MinMax scaling for neural networks
- **Early Stopping**: Prevent overfitting


```python
import sys
import warnings
warnings.filterwarnings('ignore')
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from src.data.preprocess import load_and_preprocess
from src.models.train_lstm import DemandForecaster

plt.style.use('seaborn-v0_8-darkgrid')
print("✅ Setup complete!")
```

```python
# Load data
df = load_and_preprocess()
print(f"Data shape: {df.shape}")
```

```python
# Initialize LSTM forecaster
forecaster = DemandForecaster(
    sequence_length=30,
    hidden_size=64,
    num_layers=2,
    learning_rate=0.001,
    batch_size=32,
    num_epochs=100,
    random_state=42
)

# Prepare time series data
X_train, X_test, y_train, y_test = forecaster.prepare_data(df)
```

```python
# Train LSTM model
forecaster.train_model(X_train, y_train, X_test, y_test)
```

```python
# Evaluate model
results = forecaster.evaluate_model(X_test, y_test)

# Plot training history
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(forecaster.train_losses, label='Training Loss')
ax.plot(forecaster.val_losses, label='Validation Loss')
ax.set_xlabel('Epoch')
ax.set_ylabel('MSE Loss')
ax.set_title('LSTM Training History', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Save model
forecaster.save_model()
print("\n✅ LSTM training complete!")
```

## Key Interpretation

### LSTM Model:
- Uses 30 days of history to predict next day demand
- Early stopping prevents overfitting
- MinMax scaling normalizes input for neural networks

### Evaluation Metrics:
- **RMSE**: Root Mean Square Error (lower is better)
- **MAE**: Mean Absolute Error
- **R²**: Explained variance (closer to 1 is better)
- **MAPE**: Mean Absolute Percentage Error

### Business Value:
- Enables inventory optimization
- Reduces stockouts and overstock
- Improves supply chain planning
